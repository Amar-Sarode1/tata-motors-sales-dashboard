```python
print("\nMissing values:")
print(df.isnull().sum().sort_values(ascending=False))
```

```python
print("\nDuplicate Rows:", df.duplicated().sum())
```

```python

# 1. Check incorrect dates
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
print("Invalid dates:", df["Date"].isna().sum())

# 2. Check invalid sales quantities
print("Units sold <= 0:", (df["Units_Sold"] <= 0).sum())

# 3. Check invalid prices
print("Price <= 0:", (df["Ex_Showroom_Price"] <= 0).sum())

# 4. Check invalid discounts
print("Negative discounts:", (df["Discount"] < 0).sum())
print("Discount greater than price:",
      (df["Discount"] > df["Ex_Showroom_Price"]).sum())

# 5. Check satisfaction score range (assuming 1–5)
print("Scores outside 1–5:",
      (~df["Satisfaction_Score"].between(1, 5)).sum())

# 6. Check year/month consistency
print("Year mismatch:",
      (df["Date"].dt.year != df["Year"]).sum())

print("Month mismatch:",
      (df["Date"].dt.month != df["Month"]).sum())
```

```python

# Check satisfaction score values
print("Satisfaction score values:")
print(sorted(df["Satisfaction_Score"].dropna().unique()))

print("\nSatisfaction score range:")
print(df["Satisfaction_Score"].min(),
      df["Satisfaction_Score"].max())

# Check Month column format
print("\nMonth values:")
print(df["Month"].head(10).tolist())

# Compare Date, Year, Month and Sale_Month
display(df[[
    "Date", "Year", "Month",
    "Sale_Year", "Sale_Month"
]].head(10))
```

```python

# Check how Revenue is calculated
df["Calculated_Revenue"] = (
    df["Units_Sold"] * df["Ex_Showroom_Price"]
    - df["Discount"]
)

print(
    "Revenue mismatches:",
    (~df["Revenue"].round(2).eq(
        df["Calculated_Revenue"].round(2)
    )).sum()
)

# Check whether Discount_Pct matches Discount / Price
df["Calculated_Discount_Pct"] = (
    df["Discount"] / df["Ex_Showroom_Price"] * 100
)

print(
    "Discount percentage mismatches:",
    (~df["Discount_Pct"].round(2).eq(
        df["Calculated_Discount_Pct"].round(2)
    )).sum()
)

# Check the values without modifying original columns
display(df[[
    "Units_Sold", "Ex_Showroom_Price",
    "Discount", "GST", "Revenue",
    "Discount_Pct"
]].head(10))
```

```python

# Check whether Revenue includes GST
df["Revenue_Before_GST"] = (
    df["Units_Sold"] * df["Ex_Showroom_Price"]
    - df["Discount"]
)

df["Revenue_With_GST"] = (
    df["Revenue_Before_GST"] + df["GST"]
)

print(
    "Matches before GST:",
    df["Revenue"].round(2).eq(
        df["Revenue_Before_GST"].round(2)
    ).sum()
)

print(
    "Matches after adding GST:",
    df["Revenue"].round(2).eq(
        df["Revenue_With_GST"].round(2)
    ).sum()
)

# Inspect the relationship among price, discount, GST and revenue
display(df[[
    "Units_Sold", "Ex_Showroom_Price",
    "Discount", "GST", "Revenue",
    "Revenue_Before_GST", "Revenue_With_GST"
]].head(10))
```

```python

# Compare revenue against the base vehicle amount
df["Base_Amount"] = (
    df["Units_Sold"] * df["Ex_Showroom_Price"]
)

df["Revenue_Ratio"] = df["Revenue"] / df["Base_Amount"]

print("Revenue ratio summary:")
print(df["Revenue_Ratio"].describe())

print("\nRevenue ratio examples:")
display(df[[
    "Units_Sold",
    "Ex_Showroom_Price",
    "Discount",
    "GST",
    "Revenue",
    "Revenue_Ratio"
]].head(15))

```

```python

df["Base_Amount"] = df["Units_Sold"] * df["Ex_Showroom_Price"]

df["Effective_GST_Pct"] = (
    df["GST"] / df["Base_Amount"] * 100
)

df["Revenue_Uplift_Pct"] = (
    (df["Revenue"] / df["Base_Amount"]) - 1
) * 100

print("Effective GST percentage:")
print(df["Effective_GST_Pct"].describe().round(2))

print("\nRevenue uplift percentage:")
print(df["Revenue_Uplift_Pct"].describe().round(2))

display(df[[
    "Base_Amount", "Discount", "GST",
    "Effective_GST_Pct", "Revenue_Uplift_Pct"
]].head(10))
```

```python

# Keep the original df unchanged
analysis_df = df.copy()

# Remove temporary validation columns
temp_cols = [
    "Calculated_Revenue",
    "Calculated_Discount_Pct",
    "Revenue_Before_GST",
    "Revenue_With_GST",
    "Base_Amount",
    "Revenue_Ratio",
    "Effective_GST_Pct",
    "Revenue_Uplift_Pct"
]

analysis_df = analysis_df.drop(
    columns=temp_cols,
    errors="ignore"
)

analysis_df.to_csv(
    "tata_motors_analysis.csv",
    index=False
)

print("Analysis CSV exported successfully.")
print("Shape:", analysis_df.shape)
```

```python

import pandas as pd
from sqlalchemy import create_engine, text
from getpass import getpass
from urllib.parse import quote_plus

mysql_user = input("MySQL username (usually root): ")
mysql_password = getpass("MySQL password: ")

password_encoded = quote_plus(mysql_password)

engine = create_engine(
    f"mysql+pymysql://{mysql_user}:{password_encoded}@localhost:3306/tata_motors_project"
)

with engine.connect() as conn:
    print("Connected database:", conn.execute(
        text("SELECT DATABASE()")
    ).scalar())

df_sql = pd.read_csv("tata_motors_analysis.csv")

with engine.connect() as conn:
    existing_rows = conn.execute(
        text("SELECT COUNT(*) FROM tata_sales")
    ).scalar()

print("Existing rows:", existing_rows)

if existing_rows == 0:
    df_sql.to_sql(
        name="tata_sales",
        con=engine,
        if_exists="append",
        index=False,
        chunksize=500
    )
    print("Rows sent:", len(df_sql))
else:
    print("Table already contains data. Import skipped.")

with engine.connect() as conn:
    print("Rows in MySQL:", conn.execute(
        text("SELECT COUNT(*) FROM tata_sales")
    ).scalar())
```
